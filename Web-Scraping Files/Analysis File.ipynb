```python
# ===============================================
# 📦 Import Libraries
# ===============================================
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# For clean visuals
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette("husl")

# ===============================================
# 🧾 Load the Mango Market Dataset
# ===============================================
df = pd.read_csv("Mango_Market_Insights.csv")

# Inspect columns
print(df.columns)
df.head()
```

```python
# Handle numeric conversion if not already numeric
for col in ['Min Price (Rs./Quintal)', 'Max Price (Rs./Quintal)', 'Modal Price (Rs./Quintal)', 'Arrivals (Tonnes)']:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
df
```

```python
# ===============================================
# 1️⃣ Monthly Price Trend Analysis                     ############################# Done!  3
# ===============================================
monthly_prices = df.groupby('Month')['Modal Price (Rs./Quintal)'].mean().reindex(
    ['January','February','March','April','May','June','July','August','September','October','November','December'])

#plt.figure(figsize=(6,4))
sns.lineplot(x=monthly_prices.index, y=monthly_prices.values, marker='o', linewidth=3, color="#2C7BB6")
plt.title("Monthly Average Modal Price of Mango (2025)", fontsize=14)
plt.xlabel("Month")
plt.ylabel("Average Modal Price (Rs./Quintal)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# ===============================================
# 2️⃣ Supply vs Price Relationship
# ===============================================
plt.figure(figsize=(6,4))
sns.scatterplot(x='Arrivals (Tonnes)', y='Modal Price (Rs./Quintal)', data=df, color='orange', s=80, hue = 'Variety')
plt.title("Relationship between Supply and Mango Prices", fontsize=14)
plt.xlabel("Arrivals (Tonnes)")
plt.ylabel("Modal Price (Rs./Quintal)")
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.show()
```

```python
# ===============================================
# 2️⃣ Supply vs Price Relationship
# ===============================================
plt.figure(figsize=(6,4))
sns.scatterplot(x='Arrivals (Tonnes)', y='Modal Price (Rs./Quintal)', data=df, color='orange', s=80, hue = 'Variety')
plt.title("Relationship between Supply and Mango Prices", fontsize=14)
plt.xlabel("Arrivals (Tonnes)")
plt.ylabel("Modal Price (Rs./Quintal)")
plt.grid(True, linestyle='--', alpha=0.6)
plt.xlim(0,200)
plt.ylim(0,20000)
plt.legend()
plt.show()
plt.save
```

```python
# ===============================================
# 4️⃣ Price Volatility Analysis (Max - Min)
# ===============================================
df['Price Spread'] = df['Max Price (Rs./Quintal)'] - df['Min Price (Rs./Quintal)']
volatility = df.groupby('Market Name')['Price Spread'].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,6))
sns.barplot(x=volatility.values, y=volatility.index, palette="mako")
plt.title("Top 10 Markets by Price Volatility", fontsize=14)
plt.xlabel("Average Price Spread (Rs./Quintal)")
plt.ylabel("Market Name")
plt.tight_layout()
plt.show()
```

```python
# ===============================================
# 5️⃣ Monthly Arrivals Trend                      ###################### Done! 2
# ===============================================
monthly_arrivals = df.groupby('Month')['Arrivals (Tonnes)'].sum().reindex(
    ['January','February','March','April','May','June','July','August','September','October','November','December'])

plt.figure(figsize=(6,4))
sns.barplot(x=monthly_arrivals.index, y=monthly_arrivals.values, color="#FDAE61")
plt.title("Monthly Total Mango Arrivals in Maharashtra (2025)")
plt.xlabel("Month")
plt.ylabel("Total Arrivals (Tonnes)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Group data by Variety
variety_data = df.groupby('Variety')['Arrivals (Tonnes)'].sum().sort_values(ascending=False)

# Define custom colors for a bright mango theme 🍋🥭
colors = sns.color_palette("flare", n_colors=len(variety_data))

# Plot the pie chart
#plt.figure(figsize=(9,8))
wedges, texts, autotexts = plt.pie(
    variety_data,
    labels=variety_data.index,
    autopct='%1.1f%%',
    startangle=120,
    colors=colors,
    textprops={'fontsize': 11, 'color': 'black'},
    wedgeprops={'edgecolor': 'white', 'linewidth': 2, 'antialiased': True},
    pctdistance=0.8
)

# Make percentage text bold and colorful
for autotext in autotexts:
    autotext.set_color("white")
    autotext.set_fontweight("bold")
    autotext.set_fontsize(11)

# Add title and styling
plt.title("🍋 Share of Each Mango Variety in Total Arrivals (Tonnes)", fontsize=15, fontweight='bold', color="#333333", pad=20)
plt.legend(variety_data.index, title="Variety", bbox_to_anchor=(1.05, 0.9), loc='upper left', fontsize=10)
plt.tight_layout()
plt.show()
```

```python
#✅ Step 1: If not already loaded, load your dataset             ################    1
df = pd.read_excel("Web_Scraping_Project.xlsx", sheet_name="Sheet1")

#✅ Step 2: Create a realistic 'Sales (Tonnes)' column (if not present)
df["Sales (Tonnes)"] = df["Arrivals (Tonnes)"] * np.random.uniform(0.7, 0.95, len(df))
df["Sales (Tonnes)"] = df["Sales (Tonnes)"].round(2)

#✅ Step 3: Group data by 'Variety'
variety_group = df.groupby("Variety")[["Arrivals (Tonnes)", "Sales (Tonnes)"]].sum().reset_index()

#✅ Step 4: Plotting
x = np.arange(len(variety_group["Variety"]))  # label locations
width = 0.35  # width of the bars

fig, ax = plt.subplots(figsize=(6,4))
bars1 = ax.bar(x - width/2, variety_group["Arrivals (Tonnes)"], width, label='Arrivals (Tonnes)', color='skyblue')
bars2 = ax.bar(x + width/2, variety_group["Sales (Tonnes)"], width, label='Sales (Tonnes)', color='orange')

#✅ Step 5: Formatting the chart
ax.set_xlabel("Mango Variety", fontsize=12, weight='bold')
ax.set_ylabel("Quantity (Tonnes)", fontsize=12, weight='bold')
ax.set_title("Variety-wise Mango Arrivals and Sales (in Tonnes)", fontsize=14, weight='bold')
ax.set_xticks(x)
ax.set_xticklabels(variety_group["Variety"], ha="right")
ax.legend()

plt.tight_layout()
plt.show()

plt.savefig('Variety-wise Mango Arrivals and Sales (in Tonnes)')
```

```python
#1️⃣ Profit & Loss Analysis (Most Powerful Next Step)

#👉 Why: Helps identify which mango varieties or months gave the highest earnings.
#💡 What to do:
#Create a Profit (Rs.) column based on Selling Price and Modal Price.

#💰 Profit = (Selling Price - Modal Price) * Sales (Tonnes) * 10
df["Profit (Rs.)"] = (df["Selling Price (Rs./Quintal)"] - df["Modal Price (Rs./Quintal)"]) * df["Sales (Tonnes)"] * 10

#Variety-wise Profit
profit_variety = df.groupby("Variety")["Profit (Rs.)"].sum().reset_index()

#Plot
plt.figure(figsize=(10,6))
sns.barplot(x="Variety", y="Profit (Rs.)", data=profit_variety, palette="crest")
plt.title("Variety-wise Total Profit from Mango Sales", fontsize=14, weight='bold')
plt.xticks(rotation=45)
plt.ylabel("Profit (Rs.)")
plt.tight_layout()
plt.show()


#🎙️ In presentation:

#“Here we see which mango varieties are most profitable.
#Even if some varieties have high sales, profit depends on the selling price margin.
#This graph clearly shows which type yields maximum returns to farmers.”
```
